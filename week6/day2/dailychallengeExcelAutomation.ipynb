```python
import openpyxl
from openpyxl.styles import Font, PatternFill
from openpyxl.utils.cell import get_column_letter

# 1. Create an Excel workbook using openpyxl with a sheet named “Grades”.
workbook = openpyxl.Workbook()
grades_sheet = workbook.active
grades_sheet.title = "Grades"

# 2. Populate the sheet with the data structure:
data = {
    "Joe": {
        "math": 65,
        "science": 78,
        "english": 98,
        "gym": 89
    },
    "Bill": {
        "math": 55,
        "science": 72,
        "english": 87,
        "gym": 95
    },
    "Tim": {
        "math": 100,
        "science": 45,
        "english": 75,
        "gym": 92
    },
    "Sally": {
        "math": 30,
        "science": 25,
        "english": 45,
        "gym": 100
    },
    "Jane": {
        "math": 100,
        "science": 100,
        "english": 100,
        "gym": 60
    }
}

# 3. Add a row for each student’s grades.
header = ["Student Names", "math", "science", "english", "gym"]
grades_sheet.append(header)

for student, grades in data.items():
    row_data = [student, grades["math"], grades["science"], grades["english"], grades["gym"]]
    grades_sheet.append(row_data)

# 4. Add a formula in the last row of each subject column to calculate the average grade for that subject.
num_students = len(data)
last_row = num_students + 2  # +2 because of the header row

subjects = ["math", "science", "english", "gym"]
for col_index, subject in enumerate(subjects, start=2):  # Start from column B (index 2)
    col_letter = get_column_letter(col_index)
    average_formula = f"=AVERAGE({col_letter}2:{col_letter}{last_row - 1})"
    grades_sheet[f"{col_letter}{last_row}"] = average_formula

# Add a label for the average row
grades_sheet['A' + str(last_row)] = "Average"

# 5. Apply bold and colored formatting to the header row.
header_row = grades_sheet[1]
bold_font = Font(bold=True)
header_fill = PatternFill(start_color="FFFF00", end_color="FFFF00", fill_type="solid") # Yellow

for cell in header_row:
    cell.font = bold_font
    cell.fill = header_fill

# Save the workbook to a file named gradebook.xlsx
workbook.save("gradebook.xlsx")

print("Excel gradebook 'gradebook.xlsx' has been created successfully.")
```
