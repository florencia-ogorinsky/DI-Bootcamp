```python
import openpyxl
from openpyxl.utils import column_index_from_string
from openpyxl.styles import Alignment, PatternFill
import pandas as pd
from openpyxl.utils.dataframe import dataframe_to_rows

# Create a new workbook
workbook = openpyxl.Workbook()

# Select the default active worksheet and rename it
data_sheet = workbook.active
data_sheet.title = "DataSheet"

# Create a new worksheet named "SummarySheet"
summary_sheet = workbook.create_sheet("SummarySheet")
```

```python
# Insert headers into DataSheet
data_sheet['A1'] = 'Category'
data_sheet['B1'] = 'Value 1'
data_sheet['C1'] = 'Value 2'
data_sheet['D1'] = 'Value 3'

# Add two rows of data into DataSheet
data_sheet['A2'] = 'Item A'
data_sheet['B2'] = 10
data_sheet['C2'] = 20
data_sheet['D2'] = 30

data_sheet['A3'] = 'Item B'
data_sheet['B3'] = 15
data_sheet['C3'] = 25
data_sheet['D3'] = 35
```

```python
# Merge Cells: Merge cells B1 to D1 and center-align
data_sheet.merge_cells('B1:D1')
merged_cell = data_sheet['B1']
merged_cell.value = 'Merged Header'
merged_cell.alignment = Alignment(horizontal='center')

# Color Cells: Apply a blue fill color to the merged cell
blue_fill = PatternFill(start_color="ADD8E6", end_color="ADD8E6", fill_type="solid") # Light blue
merged_cell.fill = blue_fill
```

```python
# Create a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 28],
        'City': ['New York', 'London', 'Paris']}
df = pd.DataFrame(data)

# Insert DataFrame into SummarySheet
for r_idx, row in enumerate(dataframe_to_rows(df, header=True, index=False)):
    summary_sheet.append(row)
```

```python
# Save the Workbook
workbook.save('example_workbook.xlsx')

print("Workbook 'example_workbook.xlsx' has been created and saved.")
```
